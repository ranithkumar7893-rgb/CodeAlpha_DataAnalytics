```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
electronics = [45000,48000,52000,49000,61000,58000,63000,71000,67000,72000,85000,95000]
clothing = [32000,28000,35000,38000,42000,39000,44000,47000,43000,50000,60000,70000]
groceries = [55000,53000,57000,56000,59000,61000,63000,65000,62000,64000,68000,75000]

fig, axes = plt.subplots(1, 3, figsize=(15,5))
fig.suptitle("Sales Dashboard - CodeAlpha", fontsize=14, fontweight="bold")

axes[0].plot(months, electronics, marker="o", color="blue", linewidth=2)
axes[0].set_title("Electronics Sales")
axes[0].tick_params(axis="x", rotation=45)

axes[1].bar(months, clothing, color="coral")
axes[1].set_title("Clothing Sales")
axes[1].tick_params(axis="x", rotation=45)

total = [e+c+g for e,c,g in zip(electronics,clothing,groceries)]
axes[2].pie([sum(electronics),sum(clothing),sum(groceries)],
            labels=["Electronics","Clothing","Groceries"],
            autopct="%1.1f%%", colors=["blue","coral","green"])
axes[2].set_title("Category Share")

plt.tight_layout()
plt.savefig("visualization.png")
plt.show()
print("Task 3 Done!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob

reviews = [
    "This product is absolutely amazing! Love it!",
    "Terrible quality, very disappointed.",
    "Okay product, nothing special.",
    "Excellent! Best purchase ever!",
    "Worst product I have ever bought.",
    "Good value for money, works well.",
    "Outstanding performance, highly recommend!",
    "Average product, does the job.",
    "Very poor quality, broke after 2 days.",
    "Fantastic! Will buy again!"
]

results = []
for r in reviews:
    blob = TextBlob(r)
    pol = blob.sentiment.polarity
    if pol > 0.1: sent = "Positive 😊"
    elif pol < -0.1: sent = "Negative 😞"
    else: sent = "Neutral 😐"
    results.append({"Review": r[:40], "Polarity": pol, "Sentiment": sent})

df = pd.DataFrame(results)
print(df[["Review","Sentiment"]])

counts = df["Sentiment"].value_counts()
plt.figure(figsize=(8,5))
plt.bar(counts.index, counts.values, color=["green","red","orange"])
plt.title("Sentiment Analysis - CodeAlpha")
plt.ylabel("Count")
plt.savefig("sentiment.png")
plt.show()
print("Task 4 Done!")
```
